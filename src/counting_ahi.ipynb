In this file we want to try to load a patients data and calculate their AHI based on the expert-annotated files to see if they according to AHI have sleep apnea.

```python
import numpy as np

import data_management as dm
from pathlib import Path

root_dir = Path.cwd().parent
demo_data_path = str(root_dir/"demo-data"/ "tr03-0005/tr03-0005")

#dm.extract_and_save_to_parquet(demo_data_path)
```

```python
df = dm.load_from_parquet_to_pandas(demo_data_path)

df.describe()
```

```python
# Number of apnea events for the whole night
a = df["is_apnea"].diff()
a.iloc[0] = 0 #this lets us count if we start the night with an apnea, highly unlikely

number_of_apnea = (abs(a) > 0).cumsum()

number_of_apnea
```

```python
# Number of hypopnea events for the whole night
b = df["is_hypopnea"].diff()
b.iloc[0] = 0 #this lets us count if we start the night with an hypopnea, highly unlikely

number_of_hypopnea = (abs(b) > 0).cumsum()

number_of_hypopnea
```
