```python
import scipy.io
import os

# Load the data from the mat file
data_dir = "../../data"
record = "tr03-0146"
mat_file = os.path.join(data_dir, f"{record}.mat")

mat = scipy.io.loadmat(mat_file)
vals = mat["val"]  # Shape: (channels, samples)

# Create DataFrame with each channel as a column
df = pd.DataFrame(vals.T, columns=[f"Channel_{i}" for i in range(vals.shape[0])])
print(f"Loaded {record} with shape: {df.shape}")
print(f"Columns: {df.columns.tolist()}")
df
```
